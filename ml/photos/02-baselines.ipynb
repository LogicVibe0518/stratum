```python
import pandas as pd
import photolab as lab

cases = lab.load_cases()
lab.check_cases(cases)
print(len(cases), "cases;", sum(1 for c in cases if not c.want), "where nothing fits")
print(pd.Series([c.client for c in cases]).value_counts())
```

```python
import numpy as np


def average(searcher_for_seed, seeds=200, threshold=0.5):
    runs = [lab.evaluate(cases, searcher_for_seed(s), threshold) for s in range(seeds)]
    return {key: float(np.mean([r[key] for r in runs])) for key in runs[0]}


baselines = pd.DataFrame(
    {
        "random": average(lab.random_searcher),
        "today (first photo)": lab.evaluate(cases, lab.first_photo_searcher, threshold=0.5),
    }
).T.round(3)
baselines
```
